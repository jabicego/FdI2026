1. Calculo de materiales del puente:


Dimensiones del puente: Solicitamos al usuario las dimensiones del puente

```python
##Dimensiones
Largo=float(input("Largo(m):"))
Ancho=float(input("Ancho(m):"))
Altura_Losa=float(input("Altura de la losa(m):"))
Medida_Pilares=float(input("Volúmen pilares(m3):"))
Cantidad_Pilares=int(input("Cantidad de pilares:"))
```

Volúmen de hormigón necesario: Calculamos el volúmen de horigón necesario considerando losas y pilares

```python
##Volúmen
Volúmen_hormigón_puente=Largo*Ancho*Altura_Losa
Volúmen_hormigón_Pilares=Medida_Pilares*Cantidad_Pilares
Volúmen_total=Volúmen_hormigón_puente+Volúmen_hormigón_Pilares
print("Volúmen Total hormigon:" ,Volúmen_total , "m3")
```

Cantidad acero de refuerzo: Calculamos el acero de refuerzo multiplicandolo por 150kg/m3

```python
##Acero de refuerzo
Ref_Acero=Volúmen_total*150
print("Cantidad de acero de refuerzo:", Ref_Acero , "kg")
```

Costo estimado de materiales: estimamos el costo total y parciales(Hormigón y Acero)

---



```python
##Costo estimado
Costo_Hormigón=Volúmen_total*85000
Costo_Acero=Ref_Acero*1200
Costo_Total=Costo_Hormigón+Costo_Acero
print("Costo estimado de hormigón:", Costo_Hormigón , "$")
print("Costo estimado de acero:", Costo_Acero , "$")
print("Costo estimado de materiales:", Costo_Total , "$")
```

2. Analisis de precipitaciones

```python
##Lista con datos de precipitación mensual de Villa María del último año
import csv

# Define aquí la ruta correcta de tu archivo en Drive
ruta_archivo = '/content/drive/MyDrive/Colab Notebooks/Data/datos_climaticos_villa_maria.csv'

try:
  with open(ruta_archivo) as f:
    filas=csv.reader(f)
    encabezados=next(filas) ##Separamos los encabezados
    lineas=[]
    for fila in filas:
      lineas.append(fila) ##Agregamos filas que no son encabezados

    lineas_2024=[]
    for f in lineas:
      if len(f) > 1 and f[1].startswith('2024'):
        lineas_2024.append(f) ##Limpiamos las filas que no pertenecen al último año

    mes=[]
    año=[]
    precipitaciones=[]
    temperatura=[]
    cantidad_dias_lluvia=[]

    for i in lineas_2024:
      mes.append(i[0])
      año.append(i[1])
      # Convertimos a float para poder realizar cálculos matemáticos después
      precipitaciones.append(float(i[2]))
      temperatura.append(float(i[3]))
      cantidad_dias_lluvia.append(int(i[4]))

  mes_precipitaciones=tuple(zip(mes,precipitaciones))
  print(mes_precipitaciones)

except FileNotFoundError:
  print(f"No se encontró el archivo en: {ruta_archivo}")
  print("Por favor, verifica la carpeta en tu Google Drive y actualiza la variable 'ruta_archivo'")
```

Realizamos un analisis estadístico de la resistencia de hormigón

```python
##Analisis estadistico
import csv
import math
ruta = '/content/drive/MyDrive/Colab Notebooks/Data/ensayos_suelo.csv'

with open(ruta) as f:
    filas = csv.reader(f)
    encabezados = next(filas)
    datos = list(filas)

resistencias = []
por_punto = {}

for fila in datos:
    punto = fila[0]
    resistencia = float(fila[5])
    resistencias.append(resistencia)

    if punto not in por_punto:
        por_punto[punto] = []
    por_punto[punto].append(resistencia)

## Cálculos estadísticos generales
n = len(resistencias)
media = sum(resistencias) / n
varianza = sum((x - media) ** 2 for x in resistencias) / (n - 1)
desvio_std = math.sqrt(varianza)
coef_variacion = (desvio_std / media) * 100
res_min = min(resistencias)
res_max = max(resistencias)

print("\n--- ESTADÍSTICAS GENERALES DE RESISTENCIA ---")
print(f"Media: {media:.2f} kPa | Desvío Estándar: {desvio_std:.2f} kPa | CV: {coef_variacion:.2f}%")
print(f"Mínima: {res_min:.2f} kPa | Máxima: {res_max:.2f} kPa")

## Cálculos estadísticos por punto
print("\nRESISTENCIA MEDIA Y DESVÍO POR PUNTO")
for punto, valores in por_punto.items():
    m_p = sum(valores) / len(valores)  # Media
    if len(valores) > 1:
        v_p = math.sqrt(sum((x - m_p) ** 2 for x in valores) / (len(valores) - 1))
        cv_p = (v_p / m_p) * 100  # Coeficiente de Variación
    else:
        v_p = 0.0
        cv_p = 0.0
    print(f"Punto {punto}: Media = {m_p:.2f} kPa | Desvío = {v_p:.2f} kPa | CV = {cv_p:.2f}% | Mín = {min(valores):.2f} | Máx = {max(valores):.2f}")
```
