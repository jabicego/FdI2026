```python
# 1. Lectura del archivo y conversión a lista de tuplas
archivo = open("ensayos_suelo.csv", "r", encoding="utf-8")
lineas = archivo.readlines()
archivo.close()

ensayos_tuplas = []
for linea in lineas[1:]:
    datos = linea.strip().split(",")
    if not datos or datos[0] == "":
        continue
    t = (
        datos[0],
        float(datos[1]),
        float(datos[2]),
        float(datos[3]),
        datos[4],
        float(datos[5]),
        float(datos[6]),
        int(datos[7]),
    )
    ensayos_tuplas.append(t)

# 2. Interacción con el usuario (Ingresar punto y resistencia requerida)
print("--- CONSULTA DE PROFUNDIDAD MÍNIMA ---")
punto_usuario = input("Ingrese el punto de muestreo (ej: P1): ").strip().upper()
resistencia_usuario = float(input("Ingrese el valor de resistencia requerido (kPa): "))

profundidad_minima = None
for t in ensayos_tuplas:
    if t[0] == punto_usuario and t[5] >= resistencia_usuario:
        profundidad_minima = t[3]
        break  # Encontramos la primera profundidad que cumple

if profundidad_minima is not None:
    resultado_consulta = f"En el punto {punto_usuario}, para alcanzar {resistencia_usuario} kPa se requiere una profundidad mínima de {profundidad_minima} m."
else:
    resultado_consulta = f"En el punto {punto_usuario}, ningún ensayo alcanza los {resistencia_usuario} kPa."
print(resultado_consulta)

# 3. Cálculo de la humedad promedio por cada punto de muestreo
puntos_unicos = ["P1", "P2", "P3", "P4", "P5", "P6", "P7"]
promedios_humedad = {}

for p in puntos_unicos:
    suma_humedad = 0.0
    contador = 0
    for t in ensayos_tuplas:
        if t[0] == p:
            suma_humedad += t[6]  # La humedad está en el índice 6
            contador += 1
    if contador > 0:
        promedios_humedad[p] = round(suma_humedad / contador, 2)

# 4. Guardar los resultados en un archivo de texto (informe)
informe = open("informe_suelos.txt", "w", encoding="utf-8")
informe.write("=========================================\n")
informe.write("   INFORME TÉCNICO DE SUELOS - VILLA MARÍA\n")
informe.write("=========================================\n\n")
informe.write("1. CONSULTA DE PROFUNDIDAD:\n")
informe.write("   " + resultado_consulta + "\n\n")
informe.write("2. HUMEDAD PROMEDIO POR PUNTO:\n")

for p, hum in promedios_humedad.items():
    informe.write(f"   - Punto {p}: {hum} %\n")

informe.close()
print("\n¡Informe guardado con éxito en el archivo 'informe_suelos.txt'!")
```

```python
import csv
import math
from google.colab import files

print("--- PASO 1: Sube tu archivo 'proveedores_materiales.csv' ---")
subidos = files.upload()
nombre_archivo = list(subidos.keys())[0]

# --- PASO 2: Función para buscar proveedores ---
def buscar_proveedores(material_buscado, prioridad="precio"):
    proveedores_encontrados = []

    with open(nombre_archivo, mode='r', encoding='utf-8') as archivo:
        lector = csv.reader(archivo)
        encabezado = next(lector)  # Salta la primera línea (títulos)

        for fila in lector:
            if fila:
                nombre_prov = fila[0]
                material_prov = fila[1]
                precio_prov = float(fila[2])

                if material_buscado.strip().lower() in material_prov.strip().lower():
                    proveedores_encontrados.append((nombre_prov, material_prov, precio_prov))

    # Ordenar según la prioridad elegida
    if prioridad == "precio":
        proveedores_encontrados.sort(key=lambda x: x[2])

    print(f"\nResultados para '{material_buscado}' (ordenado por {prioridad}):")
    if not proveedores_encontrados:
        print("No se encontraron proveedores para ese material.")
    else:
        for prov in proveedores_encontrados:
            print(f" - Proveedor: {prov[0]} | Material: {prov[1]} | Precio: ${prov[2]:.2f}")

    return proveedores_encontrados

# --- PASO 3: Prueba de ejecución ---
material = input('Ingrese material a solicitar: ')
buscar_proveedores(material, prioridad="precio")
```
