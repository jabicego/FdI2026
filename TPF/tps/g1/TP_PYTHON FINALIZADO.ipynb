## 1. Función de Cálculo Estructural
Definición de la función `dimensionar_zapata()` para el predimensionamiento de fundaciones superficiales (cuadradas o rectangulares) según la carga de columna y la resistencia admisible del suelo obtenida de `ensayos_suelo.csv` (con interpolación lineal para profundidades intermedias).

**Parámetros de entrada:**
* `carga_columna` (`float`): carga de la columna en $\text{kN}$.
* `punto_muestreo` (`str`): identificador del sondeo (`'P1'` a `'P7'`).
* `profundidad_deseada` (`float`): profundidad de fundación en metros.
* `forma` (`str`): `'cuadrada'` o `'rectangular'`.
* `relacion_lados` (`float`): relación largo/ancho ($1.0$ para zapata cuadrada).

```python
#Función de cálculo estructural:
#Crear una función (dimensionar_zapata()) para dimensionar una zapata según la carga y el tipo de suelo, usando datos del archivo ensayos_suelo.csv.
#Parámetros de entrada:
#carga_columna: float (kN) - carga de la columna
#punto_muestreo: string (ej: 'P1', 'P2'...'P7') - del archivo CSV
#profundidad_deseada: float (metros) - profundidad de fundación
#forma: string ('cuadrada' o 'rectangular')
#relacion_lados: float - para rectangular (largo/ancho), 1.0 para cuadrada

import math
import numpy as np
import pandas as pd

def dimensionar_zapata(
    carga_columna: float,
    punto_muestreo: str,
    profundidad_deseada: float,
    forma: str = "cuadrada",
    relacion_lados: float = 1.0,
    ruta_csv: str = "/content/ensayos_suelo (1).csv",
) -> dict:
    """Dimensiona una zapata aislada (cuadrada o rectangular) a partir de ensayos_suelo.csv."""
    forma = forma.strip().lower()
    punto_muestreo = punto_muestreo.strip().upper()

    if carga_columna <= 0:
        raise ValueError("La carga de la columna debe ser mayor a 0")

    if profundidad_deseada <= 0:
        raise ValueError("La profundidad deseada debe ser mayor a 0 m.")

    if forma == "cuadrada":
        relacion_lados = 1.0
    elif forma == "rectangular":
        if relacion_lados <= 0:
            raise ValueError("Para zapata rectangular, relacion_lados debe ser > 0.")
    else:
        raise ValueError("El parámetro 'forma' debe ser 'cuadrada' o 'rectangular'.")

    df = pd.read_csv(ruta_csv)
    df_punto = df[
        df["punto_muestreo"].str.strip().str.upper() == punto_muestreo
    ].sort_values("profundidad_m")

    if df_punto.empty:
        puntos_validos = df["punto_muestreo"].unique().tolist()
        raise ValueError(
            f"Punto '{punto_muestreo}' no encontrado. Válidos: {puntos_validos}"
        )

    prof_min = float(df_punto["profundidad_m"].min())
    prof_max = float(df_punto["profundidad_m"].max())

    if profundidad_deseada < prof_min or profundidad_deseada > prof_max:
        raise ValueError(
            f"La profundidad {profundidad_deseada}m está fuera del rango ensayado "
            f"({prof_min}m-{prof_max}m)."
        )

    fila_exacta = df_punto[np.isclose(df_punto["profundidad_m"], profundidad_deseada)]

    if not fila_exacta.empty:
        fila = fila_exacta.iloc[0]
        resistencia_kpa = float(fila["resistencia_kpa"])
        tipo_suelo = fila["tipo_suelo"]
        densidad_kg_m3 = float(fila["densidad_kg_m3"])
        humedad_pct = float(fila["humedad_porcentaje"])
        metodo_dato = "Lectura directa del ensayo"
    else:
        resistencia_kpa = float(
            np.interp(
                profundidad_deseada,
                df_punto["profundidad_m"],
                df_punto["resistencia_kpa"],
            )
        )
        densidad_kg_m3 = float(
            np.interp(
                profundidad_deseada,
                df_punto["profundidad_m"],
                df_punto["densidad_kg_m3"],
            )
        )
        humedad_pct = float(
            np.interp(
                profundidad_deseada,
                df_punto["profundidad_m"],
                df_punto["humedad_porcentaje"],
            )
        )
        estrato_sup = df_punto[df_punto["profundidad_m"] <= profundidad_deseada].iloc[-1]
        tipo_suelo = f"{estrato_sup['tipo_suelo']} (interpolado)"
        metodo_dato = "Interpolación lineal entre ensayos"

    area_requerida_m2 = carga_columna / resistencia_kpa
    ancho_teorico_m = math.sqrt(area_requerida_m2 / relacion_lados)
    largo_teorico_m = ancho_teorico_m * relacion_lados

    ancho_B_m = math.ceil(ancho_teorico_m * 20) / 20
    largo_L_m = math.ceil(largo_teorico_m * 20) / 20
    area_adoptada_m2 = round(ancho_B_m * largo_L_m, 2)
    tension_real_kpa = round(carga_columna / area_adoptada_m2, 2)

    return {
        "punto_muestreo": punto_muestreo,
        "profundidad_deseada_m": profundidad_deseada,
        "metodo_obtencion": metodo_dato,
        "tipo_suelo": tipo_suelo,
        "resistencia_suelo_kpa": round(resistencia_kpa, 2),
        "densidad_kg_m3": round(densidad_kg_m3, 1),
        "humedad_porcentaje": round(humedad_pct, 1),
        "forma": forma,
        "relacion_lados": relacion_lados,
        "carga_columna_kN": carga_columna,
        "area_requerida_m2": round(area_requerida_m2, 2),
        "ancho_B_m": ancho_B_m,
        "largo_L_m": largo_L_m,
        "area_adoptada_m2": area_adoptada_m2,
        "tension_real_kpa": tension_real_kpa,
    }
```

### 3.1 Consulta y ordenamiento de proveedores de materiales
Implementación de la función `listar_proveedores(material, criterio)` sobre `proveedores_materiales.csv` para filtrar proveedores según el insumo buscado y ordenarlos por **calidad** (descendente), **precio** unitario (ascendente) o **distancia** a obra (ascendente).

```python
import csv

ARCHIVO_PROV = 'proveedores_materiales.csv'
P_PROVEEDOR = 0
P_MATERIAL = 1
P_PRECIO = 2
P_CALIDAD = 3
P_DISTANCIA = 5
P_TELEFONO = 6

def listar_proveedores(material, criterio):
    # criterio: 'calidad', 'precio' o 'distancia'
    if criterio not in ['calidad', 'precio', 'distancia']:
        print("Criterio no válido. Usá: calidad, precio o distancia.")
        return

    candidatos = []
    with open(ARCHIVO_PROV, encoding='utf-8') as f:
        filas = csv.reader(f)
        encabezados = next(filas)
        for fila in filas:
            if material.strip().lower() in fila[P_MATERIAL].lower():
                if criterio == 'calidad':
                    valor = float(fila[P_CALIDAD])
                elif criterio == 'precio':
                    valor = float(fila[P_PRECIO])
                else:
                    valor = float(fila[P_DISTANCIA])
                candidatos.append((valor, fila[P_PROVEEDOR], fila[P_MATERIAL], fila[P_TELEFONO]))

    if len(candidatos) == 0:
        print("No hay proveedores para ese material.")
        return

    # Mejor calidad = mayor valor; mejor precio o distancia = menor valor
    candidatos.sort(reverse=(criterio == 'calidad'))

    print(f"\nProveedores de '{material}' (priorizando {criterio}):")
    for valor, proveedor, mat, telefono in candidatos:
        print(f"  {proveedor} | {mat} | Tel: {telefono} | {criterio}: {valor}")

# --- Uso interactivo ---
mat = input("Material (ej: cemento, arena, H25, acero): ")
crit = input("Priorizar (calidad / precio / distancia): ").strip().lower()
listar_proveedores(mat, crit)
```

### 3.2 Análisis estadístico de resistencia del hormigón
Procesamiento del dataset de ensayos de probetas (`concrete.csv`) mediante la función `estadisticas()` para determinar cantidad de muestras, media, desvío estándar muestral, valor mínimo y máximo:
1. **Total de probetas:** evaluación global de todas las edades de ensayo.
2. **Probetas a 28 días:** evaluación a la edad estándar de control, incluyendo el cálculo de la resistencia característica estimada ($f'_{ck} = \bar{x} - 1.65 \cdot s$).

```python
import os
import csv
import math

# --- Configuración ---
os.chdir('/content/drive/MyDrive/Colab Notebooks/LA_DATA/')
ARCHIVO = 'concrete.csv'
COL_EDAD = 7        # age (días)
COL_RESIST = 8      # strength (MPa)

# --- Función que calcula las estadísticas de una lista de números ---
def estadisticas(valores):
    n = len(valores)
    media = sum(valores) / n
    suma_cuad = 0
    for v in valores:
        suma_cuad += (v - media) ** 2
    desvio = math.sqrt(suma_cuad / (n - 1))      # desvío muestral
    return n, media, desvio, min(valores), max(valores)

# --- Leer el CSV ---
with open(ARCHIVO) as f:
    filas = csv.reader(f)
    encabezados = next(filas)
    datos = []
    for fila in filas:
        datos.append(fila)

# --- Todas las probetas ---
todas = []
for fila in datos:
    todas.append(float(fila[COL_RESIST]))

n, media, desvio, minimo, maximo = estadisticas(todas)
print("TODAS LAS PROBETAS")
print(f"  Cantidad: {n}")
print(f"  Media: {media:.2f} MPa")
print(f"  Desvío estándar: {desvio:.2f} MPa")
print(f"  Mínima: {minimo:.2f} MPa")
print(f"  Máxima: {maximo:.2f} MPa")

# --- Solo probetas a 30 días (edad estándar de ensayo) ---
a28 = []
for fila in datos:
    if float(fila[COL_EDAD]) == 28:
        a28.append(float(fila[COL_RESIST]))

n, media, desvio, minimo, maximo = estadisticas(a28)
print("\nPROBETAS A 28 DÍAS")
print(f"  Cantidad: {n}")
print(f"  Media: {media:.2f} MPa")
print(f"  Desvío estándar: {desvio:.2f} MPa")
print(f"  Mínima: {minimo:.2f} MPa")
print(f"  Máxima: {maximo:.2f} MPa")
print(f"  Resistencia característica (media - 1,65 x desvío): {media - 1.65 * desvio:.2f} MPa")
```
