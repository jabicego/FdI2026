* **Descripción del problema:** Organizar la información de las calles a intervenir y calcular el volumen de mezcla asfáltica requerida para la obra.
* **Metodología utilizada:** A partir de `calles_pavimentacion.csv`, se construyó un diccionario de Python vinculando cada calle con su longitud, se calculó la superficie total (largo por ancho) y se multiplicó por un espesor de carpeta de 5 cm (0,05 m).

```python
import pandas as pd
import math
# PARTE 2 - PUNTO 1- Función de cálculo estructural
suelos = pd.read_csv("ensayos_suelo.csv")

def dimensionar_zapata(carga_columna, punto_muestreo, profundidad_deseada, forma="cuadrada", relacion_lados=1.0):
    fila = suelos[(suelos["punto_muestreo"] == punto_muestreo) & (suelos["profundidad_m"] == profundidad_deseada)]

    if fila.empty:
        print(f"No hay datos para {punto_muestreo} a {profundidad_deseada} m.")
        return None

    resistencia_kpa = float(fila["resistencia_kpa"].mean())
    tipo_suelo = next(iter(fila["tipo_suelo"]))

    area_necesaria = carga_columna / resistencia_kpa

    if forma == "cuadrada":
        ancho = math.sqrt(area_necesaria)
        largo = ancho
    else:
        ancho = math.sqrt(area_necesaria / relacion_lados)
        largo = ancho * relacion_lados

    print(f"--- ZAPATA EN {punto_muestreo} ({profundidad_deseada} m) ---")
    print(f"Suelo: {tipo_suelo} | Resistencia: {resistencia_kpa} kPa")
    print(f"Área requerida: {area_necesaria:.2f} m2")
    print(f"Medidas ({forma}): {ancho:.2f} m x {largo:.2f} m\n")
    return ancho, largo

# Pruebas de la función
dimensionar_zapata(500, "P1", 3.0, "cuadrada")
dimensionar_zapata(800, "P6", 5.0, "rectangular", 1.5)
```

## PARTE 2: Sistema de Gestión de Proyectos

### 1. Función de cálculo estructural (`dimensionar_zapata`)
* **Descripción del problema:** Automatizar el predimensionado en planta de zapatas de fundación cuadradas o rectangulares según la carga de columna y la capacidad portante del terreno.
* **Metodología utilizada:** Se definió la función `dimensionar_zapata()` con tipado explícito (`float` y `str`) que consulta `ensayos_suelo.csv` según el punto de muestreo y profundidad, obtiene la resistencia admisible en kPa, calcula el área requerida (`Carga / Resistencia`) y despeja ancho y largo según la relación de lados.
* **Resultados obtenidos:** La función devuelve las dimensiones exactas en metros para distintos escenarios de carga y profundidad.


```python
import pandas as pd
# PARTE 2 - PUNTO 2 - Gestión de archivos de proyecto
suelos = pd.read_csv("ensayos_suelo.csv")

punto = input("Ingresá el punto de muestreo (ej: P1 a P7): ").strip().upper()
res_objetivo = float(input("Ingresá la resistencia mínima requerida (kPa): "))

datos_punto = suelos[suelos["punto_muestreo"] == punto]
cumplen = datos_punto[datos_punto["resistencia_kpa"] >= res_objetivo]

if cumplen.empty:
    resultado_prof = f"En {punto} ninguna profundidad alcanza los {res_objetivo} kPa."
else:
    prof_minima = cumplen["profundidad_m"].min()
    resultado_prof = f"En {punto} para >= {res_objetivo} kPa, la profundidad mínima es: {prof_minima} m"

print("\n" + resultado_prof)

humedad_promedio = suelos.groupby("punto_muestreo")["humedad_porcentaje"].mean()
print("\n--- HUMEDAD PROMEDIO POR PUNTO (%) ---")
print(humedad_promedio)

with open("informe_suelos_villa_maria.txt", "w", encoding="utf-8") as archivo:
    archivo.write("==================================================\n")
    archivo.write("   INFORME DE ENSAYOS DE SUELO - VILLA MARÍA\n")
    archivo.write("==================================================\n\n")
    archivo.write("1. CONSULTA DE PROFUNDIDAD MÍNIMA:\n")
    archivo.write(f"   {resultado_prof}\n\n")
    archivo.write("2. HUMEDAD PROMEDIO POR PUNTO DE MUESTREO (%):\n")
    for pt, hum in humedad_promedio.items():
        archivo.write(f"   - {pt}: {hum:.2f} %\n")

print("\nInforme guardado en 'informe_suelos_villa_maria.txt'.")
```

* **Descripción del problema:** Determinar la profundidad mínima de fundación para cumplir con una resistencia requerida, analizar la humedad del terreno por sondeo y generar un reporte exportable.
* **Metodología utilizada:** Se procesó `ensayos_suelo.csv` filtrando por punto de muestreo y resistencia mínima ingresados por teclado, se agruparon los datos con `groupby()` para obtener la humedad media por punto y se escribió un archivo de texto plano (`.txt`) con formato de informe técnico.
* **Resultados obtenidos:** Se visualizó la profundidad mínima recomendada en pantalla, los promedios de humedad de `P1` a `P7` y se generó el archivo `informe_suelos_villa_maria.txt`.


```python
import pandas as pd
# PARTE 2 - PUNTO 3 - Sistema de control de calidad

proveedores = pd.read_csv("proveedores.csv")

def buscar_proveedores(material, priorizar="precio"):
    filtrados = proveedores[proveedores["material"].str.contains(material, case=False, na=False)]

    if filtrados.empty:
        print(f"No se encontraron proveedores para '{material}'.")
        return

    if priorizar == "calidad":
        ordenados = filtrados.sort_values(by="calidad", ascending=False)
    elif priorizar == "distancia":
        ordenados = filtrados.sort_values(by="distancia_km", ascending=True)
    else:
        ordenados = filtrados.sort_values(by="precio_unitario", ascending=True)

    print(f"--- PROVEEDORES DE '{material.upper()}' (Prioridad: {priorizar}) ---")
    for _, fila in ordenados.iterrows():
        print(f"- {fila['proveedor']} | Tel: {fila['telefono']} | Precio: ${fila['precio_unitario']} | Calidad: {fila['calidad']} | Dist: {fila['distancia_km']} km")
    print()

buscar_proveedores("Hormigón H25", priorizar="precio")
buscar_proveedores("Acero ADN500", priorizar="calidad")
buscar_proveedores("Cemento", priorizar="distancia")

# Estadística de resistencia de probetas de hormigón
probetas_mpa = (26.4, 25.8, 27.1, 24.9, 28.0, 26.2, 25.5, 29.1, 26.8, 25.1, 27.5, 26.0, 24.5, 27.3, 26.9)
df_probetas = pd.DataFrame({"Resistencia_MPa": probetas_mpa})

media = df_probetas["Resistencia_MPa"].mean()
desvio = df_probetas["Resistencia_MPa"].std()
minimo = df_probetas["Resistencia_MPa"].min()
maximo = df_probetas["Resistencia_MPa"].max()
aprobadas = len(df_probetas[df_probetas["Resistencia_MPa"] >= 25.0])

print("--- ESTADÍSTICA DE ENSAYOS DE PROBETAS (H25) ---")
print(f"Total probetas: {len(probetas_mpa)}")
print(f"Resistencia media: {media:.2f} MPa")
print(f"Desvío estándar: {desvio:.2f} MPa")
print(f"Mínima: {minimo:.2f} MPa | Máxima: {maximo:.2f} MPa")
print(f"Aprobadas (>= 25 MPa): {aprobadas} de {len(probetas_mpa)}")
```
