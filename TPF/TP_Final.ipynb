## TEMÁTICA: ANÁLISIS ESTRUCTURAL Y GESTIÓN DE OBRA CIVIL EN VILLA MARÍA

### Introducción
La municipalidad de Villa María comenzó un proyecto de renovación de la infraestructura urbana que incluye:
- Construcción de un nuevo puente sobre el río Ctalamochita, en barrio Palermo.
- Repavimentación de calles
- Análisis de desagües pluviales

El equipo de ingenieros de la dirección de obras públicas de la municipalidad debe desarrollar un sistema informático para analizar y gestionar estos proyectos.

---

## ENUNCIADOS

### **CÁLCULOS BÁSICOS DE INGENIERÍA**

Desarrollar un programa que realice cálculos básicos para el proyecto de infraestructura:


1. **Cálculo de materiales del puente:**
   - Solicitar al usuario las dimensiones del puente (largo, ancho, altura de capa de losa, medidas de los pilares y cantidad). Hacer con método input.
   - Calcular volumen de hormigón necesario (considerar losa y pilares)
   - Calcular cantidad de acero de refuerzo (usar 150 kg/m³ de hormigón)
   - Mostrar el costo estimado (Hormigón: $85.000/m³, Acero: $1.200/kg)

2. **Análisis de precipitaciones:** (usar archivo "datos_climaticos_villa_maria.csv")
   - Crear una lista con los datos de precipitación mensual de Villa María del último año
   - Calcular precipitación promedio, máxima y mínima
   - Identificar meses con precipitación superior al promedio
   - Usar estructuras if/elif/else para clasificar meses según riesgo de inundación. Establecer tres umbrales, para riesgo bajo, medio y alto.

3. **Planificación de pavimentación:** (usar archivo "calles_pavimentacion.csv")
   - Crear un diccionario con calles a pavimentar y sus longitudes
   - Calcular superficie total a pavimentar
   - Determinar cantidad de asfalto necesario (espesor: 5 cm)

### **SISTEMA DE GESTIÓN DE PROYECTOS** 


1. **Funcion de cálculo estructural:**
   - Crear una función (`dimensionar_zapata()`) para dimensionar una zapata según la carga y el tipo de suelo, usando datos del archivo ensayos_suelo.csv. 
   #### Parámetros de entrada:
- `carga_columna`: float (kN) - carga de la columna
- `punto_muestreo`: string (ej: 'P1', 'P2'...'P7') - del archivo CSV
- `profundidad_deseada`: float (metros) - profundidad de fundación
- `forma`: string ('cuadrada' o 'rectangular')
- `relacion_lados`: float - para rectangular (largo/ancho), 1.0 para cuadrada

2. **Gestión de archivos de proyecto:**
   - Leer archivo CSV con datos de ensayos de suelo de Villa María 
    - Hacer un programa que permita ingresar un punto de muestreo y un valor de resistencia, y que tenga como salida el valor de profundidad mínimo que se debe alcanzar para dicha resistencia.
    - Para cada punto de muestreo, calcular la humedad promedio.
   - Guardar resultados en archivo de texto con formato de informe

4. **Sistema de control de calidad:**
   - Crear una funcion que permita ingresar un material y que el programa imprima una lista deproveedores y su número de teléfono. 
   Como parámetro de entrada, se podrá seleccionar que priorizar, mejor calidad, mejor precio unitario o menor distancia.
   - Realizar un análisis estadístico de resistencia del hormigón. Utilizar el dataset de resultados de ensayos de probetas
