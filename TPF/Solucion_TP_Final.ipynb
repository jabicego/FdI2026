```python

acero_refuerzo = 150 # cantidad de acero de refuerzo en kg/m3 de hormigon
precio_hormigon_m3 = 85000 # precio del hormigon/m2
precio_acero = 1200 # precio del acero/kg

# -------------------------------------------------------------------------------------------------------

# 1- CALCULO DEL MATERIAL DEL PUENTE

pi = 3.1416

print("1- CÁLCULO DE MATERIALES DEL PUENTE\n")

# Datos de entrada del usuario
largo_puente = float(input("Ingrese el largo del puente [m]: L = "))
ancho_puente = float(input("Ingrese el ancho del puente [m]: a = "))
altura_losa = float(input("Ingrese la altura de la capa de losa [cm]: h = "))

# Datos de entrada del usuario para los pilares
diam_pilar = float(input("Ingrese el diametro de los pilares [cm]: d = "))
largo_pilar = float(input("Ingrese el largo del pilar [m]: l = "))
num_pilares = float(input("Ingrese la cantidad de pilares: n = "))
vol_puente = largo_puente*ancho_puente*altura_losa / 100 # volumen del puente
vol_pilares = (pi*(diam_pilar/100)**2)/4 # volumen del pilote
vol_total = vol_puente + vol_pilares * num_pilares # calculo del volumen total entre losa y pilares

print(f"\n- El volumen de Hormigón necesario entre losa y pilares: {float(vol_total ):.2f} m3")

acero_refuerzo = acero_refuerzo*(vol_total) # calculo de la cantidad de acereo de refuerzo = Volumen de Hormigon * precio del acero/m3 de hormigon

print(f"\n- La cantidad de acero de refuerzo es: {round(float(acero_refuerzo),2)} kg")

Costo=acero_refuerzo*precio_acero+vol_total*precio_hormigon_m3 # costo total estimado

print(f"\n- El costo total estimado es $: {round(float(Costo),2)}")   
```

```python
# 2- ANÁLISIS DE PRECIPITACIONES ("datos_climaticos_villa_maria.csv")

import csv

# Lista con los datos de las precipitaciones
with open('datos_climaticos_villa_maria.csv') as f:
  filas = csv.reader(f)
  encabezados = next(filas)
  lineas = []
  for fila in filas:
    lineas.append(fila)


print("2- ANÁLISIS DE PRECIPITACIONES\n")
print("TABLA DE LAS PRECIPITACIONES\n") # Imprime la tabla con las precipitaciones
for linea in lineas:
    print(linea)

# Calculos estadisticos
promedio = sum(float(fila[2]) for fila in lineas) / len(lineas)
maxima = max(float(fila[2]) for fila in lineas)
minima = min(float(fila[2]) for fila in lineas)

print(f"\n- Las precipitaciones promedio fueron de: {float(promedio):.2f} mm") # muestra el resultado con 2 decimales
print(f"- Las precipitaciones maximas fueron de: {float(maxima):.2f} mm") # muestra el resultado con 2 decimales
print(f"- Las precipitaciones minimas fueron de: {float(minima):.2f} mm") # muestra el resultado con 2 decimales

# Precipitaciones superiores a la media
print(f"\n\nLos meses cuyas precipitaciones fueron superiores a la media de {promedio:.2f} mm, son:")

mes_mayor_a_promedio = []
ano_mayor_a_promedio = []
for l in lineas:
    if float(l[2])>promedio:
        ano_mayor_a_promedio.append(l[1])   # Estoy filtrando los que cumplen la condicion y guardando en mayor_a_promedio
        mes_mayor_a_promedio.append(l[0])

for i in range(len(mes_mayor_a_promedio)):
    print(f"- Mes: {mes_mayor_a_promedio[i]} del {ano_mayor_a_promedio[i]}")

# Defino los umbrales para los riesgos
umbral_bajo = 10  # mm
umbral_medio = 50 # mm
umbral_alto = 100  # mm

print()
print(f"\nLa clasificación de las precipitaciones según el riesgo de Inundación:\n")

# Evaluo cada fila para determinar el tipo de riesgo y luego se imprime en consola
for l in lineas:
    if float(l[2]) <= umbral_bajo:
        riesgo = "Bajo"
    elif float(l[2]) <= umbral_medio:
        riesgo = "Medio"
    else:
        riesgo = "Alto"

    print(f"{l[0]} del {l[1]} con {float(l[2]):.1f} mm tiene un riesgo: {riesgo}")

```

```python

# funcion para dimensionar la zapata
def dimensionar_zapata(
    datos: list,
    carga_columna: float,
    punto_muestreo: str,
    profundidad_deseada: float,
    forma: str,
    relacion_lados: float = 1.0,
):
    registro = []
    for l in datos:
        if l[0] == str(punto_muestreo) and float(l[3]) == profundidad_deseada:
            # Estoy filtrando los que cumplen con el punto de muestreo y la profundidad deseada y guardando en registro
            registro = l.copy()
    
    if len(registro) == 0:
        print(f"No se encontró un registro que cumpla con el punto de muestreo {punto_muestreo} y la profundidad deseada {profundidad_deseada}.")  
        return 
    else:
        print("El registro es: ", registro)
        
    # obtengo la resistencia del suelo (KPa es kN/m2)
    resistencia_kpa = float(registro[5])  # la resistencia está en la quinta columna

    print(f"\n- La resistencia del suelo es: {resistencia_kpa:.2f} kPa" )

    # Área mínima necesaria
    A = carga_columna / resistencia_kpa  # m²

    # Tamaños según forma
    if forma == "Cuadrada":
        lado = A**0.5
        dimensiones = {
            "lado": round(lado, 3)
        }

    elif forma == "Rectangular":

        r = relacion_lados
        B = (A / r)**0.5
        L = r * B

        dimensiones = {
            "ancho": round(float(B), 3),
            "largo": round(float(L),3)
        }

    else:
        print("La forma debe ser 'Cuadrada' o 'Rectangular'.")

    # presión real verificada
    presion = carga_columna / A

    return {"area": round(float(A), 3),
            "ancho": round(float(B), 3) if forma == "Rectangular" else None,
            "largo": round(float(L), 3) if forma == "Rectangular" else None,
            "lado": round(lado, 3) if forma == "Cuadrada" else None
    }

    
```

```python

# 2- GESTIÓN DE ARCHIVOS DE PROYECTO

print("2- GESTIÓN DE ARCHIVOS DE PROYECTO\n")

import csv

# Lista con los datos de las precipitaciones
with open('ensayos_suelo.csv') as f:
    filas = csv.reader(f)
    encabezados = next(filas)
    lineas = []

    for fila in filas:
        lineas.append(fila)

print("1- FUNCIÓN DE CÁLCULO ESTRUCTURAL\n")
print("TABLA CON DATOS DE ENSAYOS DE SUELO\n") # Corrected to print 'ensayo_suelo'

for linea in lineas:
    print(linea)

pto_muestra=input("Ingrese el punto de muestreo (Ej. P1, P2, ...): ")
carga_col=float(input("Ingrese el valor de carga de la columna (kN): "))
profundidad_deseada = float(input("Ingrese la profundidad deseada (m): "))
forma = input("Ingrese la forma de la zapata (Cuadrada o Rectangular): ")
relacion_lados = float(input("Ingrese la relación de lados (solo para zapata rectangular): "))

# Llamo a la funcion
medida_zapata = {}
medida_zapata = dimensionar_zapata(
    datos=lineas,
    carga_columna=carga_col,
    punto_muestreo=pto_muestra,
    profundidad_deseada=profundidad_deseada,
    forma=forma,
    relacion_lados=relacion_lados
)

print("El area necesaria para la zapata es: ", medida_zapata["area"])

# Humedad promedio para cada punto de muestreo

print("\nHUMEDAD PROMEDIO PARA CADA PUNTO DE MUESTREO:\n")
puntos_de_muestreo = ['P1', 'P2', 'P3', 'P4', 'P5', 'P6', 'P7']
humedad_promedio_por_punto = []
cantidad_muestras = 0
suma_humedad = 0
for punto in puntos_de_muestreo:
    for linea in lineas:
        if linea[0] == punto:
            suma_humedad += float(linea[6])
            cantidad_muestras += 1
    promedio_humedad = suma_humedad / cantidad_muestras if cantidad_muestras > 0 else 0
    humedad_promedio_por_punto.append((punto, promedio_humedad))

print("Los promedios de humedad por punto de muestreo son:\n")
for punto, promedio in humedad_promedio_por_punto:
    print(f"El promedio de los valores de Humedad para {punto} es: {promedio:.2f}%")

with open('informe.txt', 'w') as o:
    o.write("----------- Informe de análisis de humedad -----------\n\n")
    
    for punto, promedio in humedad_promedio_por_punto:
        o.write(f"El promedio de los valores de Humedad para {punto} es: {promedio:.2f}%\n")
```
